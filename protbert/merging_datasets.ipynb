```python
import polars as pl

DATASET_FILE_NAME = "datasets/dataset_merged"
DATASET_FILE_NAME_WITH_FAMILIES = "datasets/dataset_merged_with_families"

df_lehner = pl.read_csv(
    "datasets/lehner_dataset_with_sequences_normalized.csv",
)
df_lehner = df_lehner.drop_nulls(
    subset=["original_seq_full", "mut_type", "mutated_seq_full", "normalized_fitness_sigmoid"])

df_lehner = df_lehner.with_columns(
    [
        pl.lit("lehner").alias("data_source"),
    ]
)

df_lehner = df_lehner.rename({"normalized_fitness_sigmoid": "target"})

df_lehner = df_lehner.select(["original_seq_full", "mutated_seq_full", "mut_type", "target", "reverse", "data_source"])

df_lehner
```

```python
df_megascale = pl.read_csv(
    "datasets/megascale_dataset_with_ddg_normalized.csv",
)

df_megascale = df_megascale.drop_nulls(
    subset=["original_seq_full", "mut_type", "mutated_seq_full", "normalized_ddG_sigmoid"])

df_megascale = df_megascale.with_columns(
    [
        pl.lit("megascale").alias("data_source"),
    ]
)

df_megascale = df_megascale.rename({"normalized_stability": "target"})

df_megascale = df_megascale.select(
    ["original_seq_full", "mutated_seq_full", "mut_type", "target", "reverse", "data_source"])

df_megascale
```

```python
df_merged = pl.concat([df_megascale, df_lehner])
df_merged
```

```python
df_merged.write_parquet(f"{DATASET_FILE_NAME}.parquet")
```

```python
df_cath = pl.read_parquet("../classification_dataset/dataset_with_families.parquet")

lf_cath_prepared = (
    df_cath.lazy()
    .drop_nulls(subset=["cath_dominant"])
    .with_columns(
        pl.col("cath_dominant").str.split(".").alias("_split_temp")
    )
    .with_columns([
        pl.col("_split_temp").list.get(0).alias("cath_class"),  # C (Level 1)
        pl.col("_split_temp").list.get(1).alias("cath_arch"),  # A (Level 2)
        pl.col("_split_temp").list.get(2).alias("cath_topology"),  # T (Level 3)
        pl.col("_split_temp").list.get(3).alias("cath_homology"),  # H (Level 4)F
    ])
    # Vybereme jen to, co chceme poslat do joinu (včetně původního cath_dominant)
    .select([
        "original_seq_full",
        "cath_dominant",  # Původní sloupec
        "cath_all",  # Původní sloupec
        "cath_class",  # Nový
        "cath_arch",  # Nový
        "cath_topology",  # Nový
        "cath_homology"  # Nový
    ])
)
```

```python

# droping the row with null cath_dominant
query = (
    df_merged.lazy()
    .join(
        other=(
            lf_cath_prepared
            .drop_nulls(subset=["cath_dominant"])
            .select(["original_seq_full", "cath_dominant", "cath_all",
                     "cath_class", "cath_arch", "cath_topology", "cath_homology"])
            .unique(subset=["original_seq_full"], keep="first")
        ),
        on="original_seq_full",
        how="left"
    )
    # Poznámka: Pokud zde dáte drop_nulls(), chová se to ve výsledku spíše jako INNER JOIN
    # (vyhodí to sekvence, které nemají CATH záznam).
    .drop_nulls().drop_nans()
)

# 3. Spustíme výpočet a streamujeme výsledek rovnou do souboru
query.sink_parquet(f"{DATASET_FILE_NAME_WITH_FAMILIES}.parquet")

```

```python
import polars as pl
import gc

# 1. Cesta k vašemu velkému, již vytvořenému souboru
source_file = f"{DATASET_FILE_NAME_WITH_FAMILIES}.parquet"

# Definice úrovní
cath_levels = ["cath_class", "cath_arch", "cath_topology", "cath_homology"]
current_subset = []

print("Začínám paměťově úsporný výpočet...")

for level in cath_levels:
    # Přidáme aktuální úroveň do seskupení (postupně [C], [C, A], [C, A, T]...)
    current_subset.append(level)

    output_filename = f"datasets/stats_{level}.parquet"
    print(f"--- Zpracovávám úroveň: {current_subset} -> {output_filename} ---")

    # 2. Vytvoříme novou Lazy Query pro každý průchod
    # DŮLEŽITÉ: .select(current_subset) zajistí, že z disku čteme JEN to, co teď seskupujeme.
    # Ostatní sloupce se vůbec nenačtou do RAM.
    q = (
        pl.scan_parquet(source_file)
        .select(current_subset)
        .group_by(current_subset)
        .len()                # Spočítáme výskyty
        .sort("len", descending=True) # Seřadíme (to už je nad malými daty, to je OK)
    )

    # 3. ZÁPIS NA DISK (Místo .collect())
    # Toto je klíčové. Nepoužíváme collect(), ale sink_parquet().
    # Data protečou pamětí a hned zmizí na disk.
    q.sink_parquet(output_filename)

    # 4. Úklid
    print(f"Uloženo do {output_filename}")
    gc.collect() # Pro jistotu vynutíme úklid paměti v Pythonu

print("Hotovo. Všechny statistiky jsou uloženy v souborech stats_*.parquet")
```
